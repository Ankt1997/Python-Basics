 2.What is an interpreter in Python?

  ->In Python, an interpreter is a program that executes code line by line. When you run a Python program, the interpreter reads each line of code, converts it into machine code that the computer can understand, and then executes it immediately.

   ->"==" equality operator,compare the values of two objects if values are equal it returns true otherwise false.
  "is" identity operator, it compare the memory addresses of two objects if both variables refer to same object in memory it returns true,otherwise false.

  12. How does the elif statement work?

   ->The elif statement is a conditional statement that is used in conjunction with the if statement. It stands for "else if".
  13. What is the difference between for and while loops?

   ->for loops: used to iterate over a sequence of items, such as a list, tuple, string, or range of numbers.
  while loops:A while loop continues to execute as long as a specified condition is true.
  14. Describe a scenario where a while loop is more suitable than a for loop.

   ->You don't know beforehand how many times the user will enter input before providing the keyword to terminate the loop. A while loop is ideal for this indefinite iteration, as it continues to execute as long as the condition is true.
